```python
import tkinter as tk
from tkinter import messagebox
import random
import string

def generate_password():
    length = 8  # Minimum length of the password
    if password_type.get() == 'Numeric':
        characters = string.digits
    elif password_type.get() == 'Alphanumeric':
        characters = string.ascii_letters + string.digits
    elif password_type.get() == 'Alphabetic':
        characters = string.ascii_letters
    else:
        messagebox.showerror("Error", "Please select a password type.")
        return

    password = ''.join(random.choice(characters) for _ in range(length))
    password_entry.delete(0, tk.END)
    password_entry.insert(0, password)

# Create the main window
root = tk.Tk()
root.title("Password Generator")

# Password type selection
password_type = tk.StringVar()
tk.Label(root, text="Select Password Type:").pack(pady=10)
tk.Radiobutton(root, text="Numeric", variable=password_type, value="Numeric").pack(anchor=tk.W)
tk.Radiobutton(root, text="Alphanumeric", variable=password_type, value="Alphanumeric").pack(anchor=tk.W)
tk.Radiobutton(root, text="Alphabetic", variable=password_type, value="Alphabetic").pack(anchor=tk.W)

# Generate button
tk.Button(root, text="Generate Password", command=generate_password).pack(pady=10)

# Entry to show generated password
password_entry = tk.Entry(root, width=30)
password_entry.pack(pady=10)

# Start the Tkinter main loop
root.mainloop()
```
