```python
import tkinter as tk
from tkinter import filedialog, messagebox
from pytube import YouTube
import moviepy.editor as mp
import os

def download_video():
    try:
        url = entry_url.get()
        save_path = filedialog.askdirectory()
        if not save_path:
            return
        
        yt = YouTube(url)
        video = yt.streams.filter(only_audio=True).first()
        out_file = video.download(output_path=save_path)

        base, ext = os.path.splitext(out_file)
        new_file = base + '.mp3'
        os.rename(out_file, new_file)

        clip = mp.AudioFileClip(new_file)
        clip.write_audiofile(new_file)

        messagebox.showinfo("Success", "Downloaded and converted to MP3 successfully!")
    except Exception as e:
        messagebox.showerror("Error", str(e))

# Set up the GUI
root = tk.Tk()
root.title("YouTube to MP3 Downloader")

frame = tk.Frame(root)
frame.pack(padx=10, pady=10)

label_url = tk.Label(frame, text="YouTube URL:")
label_url.pack(pady=5)

entry_url = tk.Entry(frame, width=50)
entry_url.pack(pady=5)

button_download = tk.Button(frame, text="Download MP3", command=download_video)
button_download.pack(pady=10)

root.mainloop()
```

```python
import tkinter as tk
from tkinter import messagebox, filedialog
from pytube import YouTube
from pydub import AudioSegment
import os

def download_mp3():
    url = url_entry.get()
    if not url:
        messagebox.showerror("Error", "Please enter a YouTube URL")
        return
    
    try:
        yt = YouTube(url)
        stream = yt.streams.filter(only_audio=True).first()
        out_file = stream.download(output_path=download_path.get())
        
        base, ext = os.path.splitext(out_file)
        new_file = base + '.mp3'
        audio = AudioSegment.from_file(out_file)
        audio.export(new_file, format="mp3")
        
        os.remove(out_file)
        
        messagebox.showinfo("Success", f"Downloaded and converted to MP3: {new_file}")
    except Exception as e:
        messagebox.showerror("Error", str(e))

def browse_directory():
    download_dir = filedialog.askdirectory()
    if download_dir:
        download_path.set(download_dir)

app = tk.Tk()
app.title("YouTube to MP3 Downloader")

url_label = tk.Label(app, text="YouTube URL:")
url_label.grid(row=0, column=0, padx=10, pady=10)

url_entry = tk.Entry(app, width=50)
url_entry.grid(row=0, column=1, padx=10, pady=10)

download_path = tk.StringVar()
path_label = tk.Label(app, text="Download Path:")
path_label.grid(row=1, column=0, padx=10, pady=10)

path_entry = tk.Entry(app, textvariable=download_path, width=38)
path_entry.grid(row=1, column=1, padx=10, pady=10, sticky='w')

browse_button = tk.Button(app, text="Browse", command=browse_directory)
browse_button.grid(row=1, column=1, padx=10, pady=10, sticky='e')

download_button = tk.Button(app, text="Download", command=download_mp3)
download_button.grid(row=2, columnspan=2, pady=20)

app.mainloop()
```

```python
from tkinter import *
from tkinter import ttk
from pytube import YouTube
from tkinter.messagebox import showinfo, showerror, askokcancel
import threading
import os
class YouTubeMP3Downloader:
    def __init__(self, root):
        self.root = root
        self.root.title("YouTube MP3 Downloader")

        # Create widgets
        self.url_label = Label(root, text="Enter YouTube URL:")
        self.url_entry = Entry(root, width=40)
        self.download_button = Button(root, text="Download MP3", command=self.download_mp3)
        self.progress_bar = ttk.Progressbar(root, mode="indeterminate")

        # Grid layout
        self.url_label.grid(row=0, column=0, padx=10, pady=10)
        self.url_entry.grid(row=0, column=1, padx=10, pady=10)
        self.download_button.grid(row=1, column=0, columnspan=2, padx=10, pady=10)
        self.progress_bar.grid(row=2, column=0, columnspan=2, padx=10, pady=10)

    def download_mp3(self):
        url = self.url_entry.get()
        try:
            yt = YouTube(url)
            audio_stream = yt.streams.filter(only_audio=True).first()
            output_path = os.path.expanduser("~")  # Change this to your desired output directory
            filename = "audio.mp3"
            audio_stream.download(output_path=output_path, filename=filename)
            showinfo("Success", f"Audio downloaded to {output_path}/{filename}")
        except Exception as e:
            showerror("Error", f"An error occurred: {str(e)}")

if __name__ == "__main__":
    root = Tk()
    app = YouTubeMP3Downloader(root)
    root.mainloop()
```
