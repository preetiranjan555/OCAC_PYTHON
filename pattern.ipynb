```python
import turtle

def draw_pyramid(levels, length):
    turtle.speed(1)
    turtle.penup()
    turtle.goto(-length * levels // 2, -length * levels // 2)
    turtle.pendown()

    for i in range(levels):
        for j in range(i + 1):
            turtle.penup()
            turtle.goto(j * length - i * length // 2, i * length)
            turtle.pendown()
            for _ in range(4):
                turtle.forward(length)
                turtle.right(90)

    turtle.done()

# Draw a pyramid with 5 levels, each block having a length of 30
draw_pyramid(5, 30)
```

```python
# Define the height of the pyramid
height = 5

# Loop to iterate through the rows
for i in range(1, height + 1):
    # Print spaces for the current row
    for j in range(height - i):
        print(" ", end="")
    # Print stars for the current row
    for k in range(2 * i - 1):
        print("*", end="")
    # Move to the next line after each row is printed
    print()
```

```python
# Number of levels in the pyramid
n = 5

# Outer loop for the number of levels
for i in range(n):
    # Inner loop for spaces
    for j in range(n-i-1):
        print(" ", end="")
    # Inner loop for stars
    for k in range(2*i+1):
        print("*", end="")
    # New line after each level
    print()
```
