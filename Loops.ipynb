```python
def fact():
    n=int(input("enter n"))
    if n<0:
        print("no result")
    elif n==0:
        print("factorial is 1")
    elif n>=1:
        fact=1
    for i in range(1,n+1):
        fact*=i
    print("The factorial is:",fact)
```
