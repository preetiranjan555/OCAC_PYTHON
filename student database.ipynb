```python
import re

def add_student(database):
    while True:
        student_id = input("Enter student ID (alphanumeric): ")
        if student_id.isalnum():
            break
        else:
            print("Invalid ID. Please enter an alphanumeric ID.")

    while True:
        name = input("Enter student name (alphabets only): ")
        if name.isalpha():
            break
        else:
            print("Invalid name. Please enter a name with alphabetic characters only.")
    
    grade = input("Enter student grade: ")
    roll_no = input("Enter student roll number: ")

    # Collect subject marks
    marks = {}
    marks["AI"] = input("Enter marks for AI: ")
    marks["ML"] = input("Enter marks for ML: ")
    marks["Science"] = input("Enter marks for Science: ")
    marks["English"] = input("Enter marks for English: ")
    marks["Python"] = input("Enter marks for Python: ")

    # Create a dictionary for the student
    student_info = {
        "name": name,
        "grade": grade,
        "roll_no": roll_no,
        "marks": marks
    }
    
    # Add the student to the database
    database[student_id] = student_info
    print("Student added successfully!\n")

def view_all_students(database):
    if not database:
        print("No students in the database.\n")
        return

    for student_id, student_info in database.items():
        print(f"ID: {student_id}")
        print(f"Name: {student_info['name']}")
        print(f"Grade: {student_info['grade']}")
        print(f"Roll No: {student_info['roll_no']}")
        print("Marks:")
        for subject, mark in student_info["marks"].items():
            print(f"  {subject}: {mark}")
        print("\n")

def view_student_by_id(database):
    student_id = input("Enter the student ID to view details: ")
    student_info = database.get(student_id)
    
    if student_info:
        print(f"ID: {student_id}")
        print(f"Name: {student_info['name']}")
        print(f"Grade: {student_info['grade']}")
        print(f"Roll No: {student_info['roll_no']}")
        print("Marks:")
        for subject, mark in student_info["marks"].items():
            print(f"  {subject}: {mark}")
        print("\n")
    else:
        print(f"No student found with ID: {student_id}\n")

def main():
    student_database = {}
    while True:
        print("1. Add student")
        print("2. View all students")
        print("3. View student by ID")
        print("4. Exit")
        
        choice = input("Enter your choice: ")
        
        if choice == "1":
            add_student(student_database)
        elif choice == "2":
            view_all_students(student_database)
        elif choice == "3":
            view_student_by_id(student_database)
        elif choice == "4":
            print("Exiting...")
            break
        else:
            print("Invalid choice. Please try again.\n")

if _name_ == "_main_":
    main()
```
